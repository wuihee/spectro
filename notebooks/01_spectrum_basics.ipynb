# 01 · The `Spectrum` Object

How to create a `Spectrum` and plot it.

1. Create a spectrum
2. Plot a spectrum

Code: `src/spectro/spectrum.py`, `plotting.py`.

## Setup

```python
import numpy as np

from spectro import Spectrum, plot_spectrum
```

## 1. Create a Spectrum

Pass wavenumbers (`x`, in cm⁻¹) and signal values (`y`). Here `y` is a made-up IR absorbance spectrum with an O–H band, two C–H bands and a C=O band.

```python
x = np.linspace(400, 4000, 1800)


def band(center, width, height):
    return height * np.exp(-0.5 * ((x - center) / width) ** 2)


y = (
    band(3350, 150, 0.6)  # O–H
    + band(2950, 20, 0.4)  # C–H
    + band(2870, 15, 0.3)  # C–H
    + band(1715, 12, 0.9)  # C=O
)

spectrum = Spectrum(x=x, y=y)
```

The data and units are available as attributes:

```python
print(spectrum.x[:3], spectrum.x_unit)
print(spectrum.y[:3], spectrum.y_unit)
```

## 2. Plot a Spectrum

`plot_spectrum` labels the axes from the units and puts high wavenumber on the left, as is usual for IR.

```python
plot_spectrum(spectrum);
```
